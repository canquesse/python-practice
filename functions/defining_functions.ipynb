# Defining Functions
A short Python fundamentals exercise. Run the cells in order.

```python
def greeting(name):
    print("Welcome, " + name)
```

```python
greeting("Can")
```

```python
greeting("Ada")
```

```python
def greeting2(name, department):
    print("Welcome, " + name)
    print("You are part of " + department)
```

```python
greeting2("Can", "Data Science Department")
```
